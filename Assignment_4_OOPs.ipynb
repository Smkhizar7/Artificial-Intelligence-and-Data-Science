1. Create a class 'BankAccount' with attributes account holder name, account number, and
balance. Add methods for deposit and withdrawal. Implement logic to prevent withdrawal if balance
is insufficient.

```python
class BankAccount:
    def __init__(self,account_holder_name,account_number,balance):
        # Attributes
        self.account_holder_name = account_holder_name
        self.account_number = account_number 
        self.balance = balance

    def deposit(self,amount):
        self.balance += amount
        print(f"You deposited {amount} in your bank account. Your balance is now {self.balance}.")

    def withdrawal(self,amount):
        if self.balance < amount:
            print("You have insufficient balance.")
        else:
            self.balance -= amount
            print(f"You withdraw {amount} from your bank account. Your balance is now {self.balance}.")
```

8. Build a class 'ShoppingCart' where user can add items with price. Implement logic to calculate
total bill and apply discount if total exceeds a certain amount.

```python
class ShoppingCart:
    def __init__(self):
        self.cart = []

    def addItem(self,item_name,item_price):
        self.cart.append({"name":item_name,"price":item_price})

    def applyDiscount(self,total):
        if total >= 1000:
            return total-total*0.05
        else:
            return total

    def calculateTotal(self):
        total_price = 0
        for item in self.cart:
            total_price += item["price"] 
        return self.applyDiscount(total_price)
```

10. Design a class 'Library' where users can borrow and return books. Add logic to check
availability before issuing a book.

```python
class Library:
    def __init__(self):
        self.books = []

    def addBook(self,book_name,book_availability):
        self.books.append({"name":book_name,"availability":book_availability})
        return "Book added"
    
    def borrowBook(self,book_name):
        selectedBook = None
        for book in self.books:
            if book["name"] == book_name:
                selectedBook = book
        if not selectedBook:
            return "Book not found"
        if not selectedBook["availability"]:
            return "Book not available"
        selectedBook["availability"] = False
        return f"You borrowed {selectedBook["name"]} book"
        
    def returnBook(self,book_name):
        for book in self.books:
            if book["name"] == book_name:
                book["availability"] = True
        return f"You returned {book_name} book"
```

```python
library1 = Library()
print(library1.addBook("Rich Dad Poor Dad",True))
print(library1.borrowBook("Rich Dad Poor Dad"))
print(library1.borrowBook("Rich Dad Poor Dad"))
print(library1.returnBook("Rich Dad Poor Dad"))
```

```python
import uuid
from abc import ABC, abstractmethod

class Flight(ABC):

    def __init__(self, flight_number, destination, total_seats, base_price):
        self.flight_number = flight_number
        self.destination = destination
        self.total_seats = total_seats
        self.available_seats = total_seats
        self.__base_price = base_price

    def get_base_price(self):
        return self.__base_price

    def set_base_price(self, new_price):
        if new_price > 0:
            self.__base_price = new_price

    def reserve_seat(self):
        if self.available_seats > 0:
            self.available_seats -= 1
            return True
        return False

    @abstractmethod
    def calculate_final_fare(self, passenger_age):
        """Abstract method to be implemented by flight subclasses"""
        pass


class EconomyFlight(Flight):

    def calculate_final_fare(self, passenger_age):
        fare = self.get_base_price() * 1.0
        return self.apply_modifiers(fare, passenger_age)

    def apply_modifiers(self, fare, age):
        if age < 12:
            fare *= 0.75
        elif age >= 60:
            fare *= 0.85

        fare *= 1.12
        return round(fare, 2)


class BusinessFlight(Flight):

    def calculate_final_fare(self, passenger_age):
        fare = self.get_base_price() * 2.5
        return self.apply_modifiers(fare, passenger_age)

    def apply_modifiers(self, fare, age):
        if age < 12:
            fare *= 0.85
        elif age >= 60:
            fare *= 0.90

        fare *= 1.12
        return round(fare, 2)

class FirstClassFlight(Flight):

    def calculate_final_fare(self, passenger_age):
        fare = self.get_base_price() * 4.0
        return self.apply_modifiers(fare, passenger_age)

    def apply_modifiers(self, fare, age):
        if age < 12 or age >= 60:
            fare *= 0.95

        fare *= 1.12
        return round(fare, 2)

class BookingSystem:

    def __init__(self, passenger_name, age, flight_instance):
        self.passenger_name = passenger_name
        self.age = age
        self.flight = flight_instance
        self.booking_id = f"TX-{uuid.uuid4().hex[:8].upper()}"
        self.final_fare = self.flight.calculate_final_fare(self.age)

    def process_payment(self, payment_method, plan_type="full"):
        receipt = f"\n=== RECEIPT (ID: {self.booking_id}) ===\n"
        receipt += f"Passenger: {self.passenger_name} (Age: {self.age})\n"
        receipt += f"Flight: {self.flight.flight_number} to {self.flight.destination}\n"
        receipt += f"Class Type: {type(self.flight).__name__}\n"
        receipt += f"Total Fare (Inc. 12% Tax & Discounts): ${self.final_fare}\n"
        receipt += f"Payment Method: {payment_method.upper()}\n"

        if plan_type.lower() == "installment":
            # Breakdown payment into 3 installments
            monthly_installment = round(self.final_fare / 3, 2)
            receipt += f"Payment Schedule: 3-Month Installment Plan\n"
            receipt += f"Initial Payment Charged: ${monthly_installment}\n"
            receipt += f"Remaining Balance: 2 payments of ${monthly_installment}/mo\n"
        else:
            receipt += f"Payment Status: Paid In Full (${self.final_fare})\n"

        receipt += "====================================\n"
        return receipt

    def save_ticket_to_file(self, receipt_data):
        filename = f"booking_{self.booking_id}.txt"
        with open(filename, "w") as file:
            file.write(receipt_data)
        print(f"Ticket securely saved to local file: '{filename}'")
```

```python
economy_flight = EconomyFlight("PK-301", "London", total_seats=2, base_price=500)
business_flight = BusinessFlight("PK-786", "New York", total_seats=2, base_price=1200)

booking1 = BookingSystem(passenger_name="Alice Smith", age=65, flight_instance=economy_flight)
receipt1 = booking1.process_payment(payment_method="Credit Card", plan_type="installment")
print(receipt1)
booking1.save_ticket_to_file(receipt1)

booking2 = BookingSystem(passenger_name="Leo Junior", age=8, flight_instance=business_flight)
receipt2 = booking2.process_payment(payment_method="Digital Wallet", plan_type="full")
print(receipt2)
booking2.save_ticket_to_file(receipt2)
```

```python
import uuid
from abc import ABC, abstractmethod


class Product(ABC):

    def __init__(self, name, stock, initial_price):
        self.name = name
        self.stock = stock
        self.__price = initial_price

    def get_price(self):
        return self.__price

    def set_price(self, new_price):
        if new_price > 0:
            self.__price = new_price

    def reduce_stock(self, quantity=1):
        if self.stock >= quantity:
            self.stock -= quantity
            return True
        return False

    @abstractmethod
    def get_eligible_discount(self):
        pass


class AC(Product):

    def get_eligible_discount(self):
        return 0.10 if self.get_price() > 800 else 0.05


class Refrigerator(Product):

    def get_eligible_discount(self):
        return 0.08


class WashingMachine(Product):

    def get_eligible_discount(self):
        return 0.12 if self.stock > 5 else 0.03


class Customer:

    def __init__(self, name):
        self.name = name
        self.cart = []

    def add_to_cart(self, product, quantity=1):
        if product.stock == 0:
            return False
        elif product.stock < quantity:
            return False
        else:
            product.reduce_stock(quantity)
            self.cart.append((product, quantity))
            return True


class OrderProcessor:

    def __init__(self, customer):
        self.customer = customer
        self.invoice_id = f"INV-{uuid.uuid4().hex[:8].upper()}"

    def calculate_totals(self):
        subtotal = 0.0
        total_discount = 0.0

        for product, qty in self.customer.cart:
            item_price = product.get_price() * qty
            discount_rate = product.get_eligible_discount()

            subtotal += item_price
            total_discount += item_price * discount_rate

        taxable_amount = subtotal - total_discount
        tax_amount = taxable_amount * 0.18
        final_total = taxable_amount + tax_amount

        return round(subtotal, 2), round(total_discount, 2), round(tax_amount, 2), round(final_total, 2)

    def generate_invoice(self, payment_method, plan_type="full"):
        if not self.customer.cart:
            return

        subtotal, discount, tax, final_total = self.calculate_totals()

        invoice_text = f"Invoice ID: {self.invoice_id}\n"
        invoice_text += f"Customer Name: {self.customer.name}\n"
        for product, qty in self.customer.cart:
            invoice_text += f"- {product.name} x{qty}: {product.get_price() * qty}\n"
        invoice_text += f"Subtotal: {subtotal}\n"
        invoice_text += f"Discount Applied: {discount}\n"
        invoice_text += f"Sales Tax (18%): {tax}\n"
        invoice_text += f"Grand Total: {final_total}\n"
        invoice_text += f"Payment Method: {payment_method}\n"

        if plan_type.lower() == "installment":
            monthly_payment = round(final_total / 6, 2)
            invoice_text += f"Payment Mode: 6-Month Installment Plan\n"
            invoice_text += f"Monthly Installment: {monthly_payment}\n"
        else:
            invoice_text += f"Payment Mode: Full Payment\n"

        self.save_to_file(invoice_text)
        return invoice_text

    def save_to_file(self, content):
        filename = f"invoice_{self.invoice_id}.txt"
        with open(filename, "w") as file:
            file.write(content)
```

```python
import uuid
from abc import ABC, abstractmethod


class FoodItem(ABC):

    def __init__(self, name, price, quantity, is_available=True):
        self.name = name
        self.__price = price
        self.__quantity = quantity
        self.is_available = is_available

    def get_price(self):
        return self.__price

    def set_price(self, price):
        if price >= 0:
            self.__price = price

    def get_quantity(self):
        return self.__quantity

    def set_quantity(self, quantity):
        if quantity >= 0:
            self.__quantity = quantity

    @abstractmethod
    def get_category(self):
        pass


class Pizza(FoodItem):

    def get_category(self):
        return "Pizza"


class Burger(FoodItem):

    def get_category(self):
        return "Burger"


class Drink(FoodItem):

    def get_category(self):
        return "Drink"


class Customer:

    def __init__(self, name):
        self.name = name
        self.cart = []

    def add_to_cart(self, item, qty):
        if not item.is_available:
            return False
        if item.get_quantity() < qty:
            return False

        item.set_quantity(item.get_quantity() - qty)
        self.cart.append((item, qty))
        return True


class OrderSystem:

    def __init__(self, customer):
        self.customer = customer
        self.order_id = f"ORD-{uuid.uuid4().hex[:8].upper()}"

    def calculate_bill(self):
        subtotal = sum(item.get_price() * qty for item, qty in self.customer.cart)

        discount = 0.0
        if subtotal > 2000:
            discount = subtotal * 0.10

        taxable_amount = subtotal - discount
        tax = taxable_amount * 0.15
        total = taxable_amount + tax

        return subtotal, discount, tax, total

    def generate_invoice(self, payment_method):
        if not self.customer.cart:
            return

        subtotal, discount, tax, total = self.calculate_bill()

        invoice = f"Order ID: {self.order_id}\n"
        invoice += f"Customer: {self.customer.name}\n"
        invoice += "Items Ordered:\n"
        for item, qty in self.customer.cart:
            invoice += f"- {item.name} ({item.get_category()}) x{qty}: {item.get_price() * qty}\n"
        invoice += f"Subtotal: {subtotal}\n"
        invoice += f"Discount (10%): {discount}\n"
        invoice += f"Tax (15%): {tax}\n"
        invoice += f"Total Bill: {total}\n"
        invoice += f"Payment Method: {payment_method}\n"

        self.save_to_file(invoice)
        return invoice

    def save_to_file(self, invoice_data):
        with open(f"order_{self.order_id}.txt", "w") as file:
            file.write(invoice_data)
```

```python
pizza_item = Pizza("Pepperoni Pizza", 1200, 10, True)
burger_item = Burger("Zinger Burger", 500, 5, True)
drink_item = Drink("Cola", 150, 20, False)

cust = Customer("Alex")
cust.add_to_cart(pizza_item, 2)
cust.add_to_cart(burger_item, 1)
cust.add_to_cart(drink_item, 2)

order = OrderSystem(cust)
res = order.generate_invoice("Card")
```
